```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression,SGDRegressor
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import  CountVectorizer
from sklearn.ensemble import (RandomForestClassifier,RandomForestRegressor,HistGradientBoostingClassifier)
from sklearn.naive_bayes import (BernoulliNB,MultinomialNB,GaussianNB)
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from xgboost import XGBClassifier


```

All possible Libraries/Functionalities are imported so that you can also try any of them and make this model more perfect. I started working with the following Classes and Functionalitites:
1) LabelEncoder() : Sentiments(target variable) in dataset are in categorical format with help of this we can convert it in form of some  numerical repressentation.

2) CountVectorizer(): To extract some meaningfull unique words and convert it in numerical format so that model can identify on the basis of these words that whether a review is positive or negative.

3) MultinomialNB(): Classification Naives Bayes model which I select to train . You can try and select different Models.

4) Joblib(): This helps to create a file of model created from above functionalitites and can can be used in any other file.

Loading and Transforming data 

```python
Ibm=pd.read_csv('E:/pyhton/hamza/IMDB Dataset.csv')
Ibm['sentiment']=LabelEncoder.fit_transform(Ibm,Ibm['sentiment'])
x=Ibm['review']
y=Ibm['sentiment']
cv=CountVectorizer()
transformed_data=cv.fit_transform(x)
X_train, X_test, y_train, y_test =train_test_split(transformed_data,y,test_size=0.2,random_state=42)
```

Model Selection and Fit on Train X and Y

```python
Model=MultinomialNB()
Model.fit(X_train,y_train)
review=['story is good I give it 4 star ']
Review=cv.transform(review)
for i in  Model.predict(Review):
    if i ==0:
        print('Negative Review')
    else:
        print('Positive Review')
```

Converting into Joblib File 

```python
import joblib 
jb=joblib
jb.dump((cv,Model),'Movie_review_prediction_model.joblib')
```
